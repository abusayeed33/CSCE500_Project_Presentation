```python
def manhattan(p1, p2):
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])

def find_closest_pair(points):
    min_dist = float('inf')
    closest_pair = (-1, -1)

    # Define 4 transformations
    transforms = [
        lambda x, y: x + y,
        lambda x, y: x - y,
        lambda x, y: -x + y,
        lambda x, y: -x - y
    ]

    for transform in transforms:
        # Sort points by the transformation
        transformed = sorted([(transform(x, y), x, y, i) for i, (x, y) in enumerate(points)], key=lambda t: t[0])

        # Check only adjacent pairs
        for i in range(1, len(transformed)):
            _, x1, y1, idx1 = transformed[i - 1]
            _, x2, y2, idx2 = transformed[i]
            dist = manhattan((x1, y1), (x2, y2))
            if dist < min_dist:
                min_dist = dist
                closest_pair = (idx1, idx2)

    # Output 1-based indices and sorted
    i1, i2 = sorted([closest_pair[0] + 1, closest_pair[1] + 1])
    return min_dist, (i1, i2)

# Example input
points = [
    (0, 0),
    (1, 2),
    (5, 5),
    (-2, -3),
    (4, 6),
    (7, 8),
    (9, 2)
]

# Run the algorithm
dist, (i1, i2) = find_closest_pair(points)

# Print output
print(dist)
print(i1, i2)
```
