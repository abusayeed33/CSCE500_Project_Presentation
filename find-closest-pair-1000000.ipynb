```python
import time

def manhattan(p1, p2):
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])

def get_transformed_lists(points):
    transformed_lists = [[] for _ in range(4)]
    for idx, (x, y) in enumerate(points, 1):
        transformed_lists[0].append((x + y, x, y, idx))
        transformed_lists[1].append((x - y, x, y, idx))
        transformed_lists[2].append((-x + y, x, y, idx))
        transformed_lists[3].append((-x - y, x, y, idx))
    return transformed_lists

def find_closest_pair(points):
    min_dist = float('inf')
    min_pair = (-1, -1)

    transformed_sets = get_transformed_lists(points)

    for transformed in transformed_sets:
        transformed.sort(key=lambda t: t[0])  # Sort by transformed value
        for i in range(1, len(transformed)):
            _, x1, y1, idx1 = transformed[i - 1]
            _, x2, y2, idx2 = transformed[i]
            dist = manhattan((x1, y1), (x2, y2))
            if dist < min_dist:
                min_dist = dist
                min_pair = (idx1, idx2)

    return min_dist, min_pair

# ✅ File-based version for Jupyter or script usage
def run_from_file(filename="/kaggle/input/input-txt/input.txt"):
    with open(filename, "r") as f:
        start_time = time.time()

        n = int(f.readline())
        points = [tuple(map(int, f.readline().split())) for _ in range(n)]

        dist, (i1, i2) = find_closest_pair(points)

        print(dist)
        print(f"{i1} {i2}")

        end_time = time.time()
        print(f"Execution time: {end_time - start_time:.4f} seconds")

# 🟢 Run this line to execute in Jupyter
run_from_file("/kaggle/input/input-txt/input.txt")
```
